```python
import pandas as pd
df= pd.read_csv('../data_sets/credito.csv')
```

```python
(df == 'na').sum()
```

```python
df.replace(["na", "NA", "N/A", "null"], pd.NA, inplace=True)
df.isna().sum()
```

```python
df.dtypes
```

```python
df.head()
```

```python
df.estado_civil.mode()
```

```python
df['escolaridade'] = df['escolaridade'].fillna(df['escolaridade'].mode)
df['estado_civil'] = df['estado_civil'].fillna(df['estado_civil'].mode)
df['salario_anual'] = df['salario_anual'].fillna(df['salario_anual'].mode)
df.isna().sum()
```

```python
df.head()
```

```python
df['limite_credito'] = df['limite_credito'].str.replace({'.':'', ',': '.'}) 
df['valor_transacoes_12m'] = df['valor_transacoes_12m'].str.replace({'.':'', ',': '.'}) 
df['salario_anual'] = df['salario_anual'].str.replace({'$':'', 'K': '000', 'menos que': '0 -'}) 
```

```python
df.head()
```

```python
df['salario_anual'] = df['salario_anual'].fillna(df['salario_anual'].mode)
df.isna().sum()
```

```python
df.to_excel('../data_sets/datos_limpios.xlsx', index= False, sheet_name='hoja_1')
```
