$ Explorar y guardas data_set limpio, con nulos para el dashboard

```python
import pandas as pd
df = pd.read_csv('../data_sets/credito.csv')
```

```python
df.dtypes
```

- ver las el n veces que aparece un X valor en las columnas (sirve para filtrado)

```python
(df == "na").sum()
```

- Pasar valor a nulos reales

```python
df.replace(["na", "NA", "N/A", "null"], pd.NA, inplace=True)
df.isna().sum()
```

```python
df.head(2)
```

```python
df['salario_anual'] = df['salario_anual'].str.replace('K', '000')
df['salario_anual'] = df['salario_anual'].str.replace('$', '')
df.head(10)
```

```python
df.sexo.str.upper
```

```python
df.dtypes
```

```python
df['valor_transacoes_12m'] = df['valor_transacoes_12m'].str.replace('.', '')
df['valor_transacoes_12m'] = df['valor_transacoes_12m'].str.replace(',', '.')
```

```python
df['limite_credito'] = df['limite_credito'].str.replace('.', '')
df['limite_credito'] = df['limite_credito'].str.replace(',', '.')
```

```python
df.head(5)
```

```python
df['valor_transacoes_12m'] = df['valor_transacoes_12m'].astype(float)
df['limite_credito'] = df['limite_credito'].astype(float)
```

```python
df.dtypes
```

```python
df.to_excel('..\data_Sets\datos_limpios.xlsx', index=False, sheet_name='hoja_1')
```
